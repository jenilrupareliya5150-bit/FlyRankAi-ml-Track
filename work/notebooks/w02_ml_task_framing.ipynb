## 1. My lane as an ML task (type)


I selected **Refresh Opportunity Scoring** because the business decision is to prioritize webpages for content updates rather than simply deciding whether a page should or should not be refreshed.

I considered other ML task types before selecting scoring:

- **Classification** was not chosen because it would only predict categories such as "Refresh" or "Do Not Refresh". It would not prioritize pages when many pages are predicted as refresh candidates.
- **Clustering** was not chosen because grouping similar pages does not answer the business question of which page should be refreshed first.
- **Ranking** is closely related, but the model first assigns every page a refresh opportunity **score**. The SEO team can then sort pages by this score to create the final ranking.

Therefore, the most appropriate ML framing for this lane is **Scoring**, where every webpage receives a numerical priority score indicating its refresh opportunity.


```python
lane = "Refresh Opportunity Scoring"
ml_task = "Scoring"

print(f"Selected Lane : {lane}")
print(f"ML Task       : {ml_task}")

comparison = {
    "Classification": "Predicts labels only",
    "Clustering": "Groups similar pages",
    "Ranking": "Orders pages using scores",
    "Scoring": "Assigns a priority score to every page"
}

for task, purpose in comparison.items():
    print(f"{task:15} -> {purpose}")
```

## 2. Target or proxy

The business goal is to identify webpages that are strong candidates for a content refresh. However, the dataset does not contain a direct label such as **"Needs Refresh"**, so a supervised model cannot learn from that label directly.

Instead, the model relies on a **proxy target** derived from observed historical SEO performance. Signals such as impressions, clicks, CTR, average position, and search trends provide measurable evidence of how a page has performed over time. These observed outcomes approximate refresh opportunity without relying on manually defined rules.

Therefore, the target is represented indirectly through historical performance signals rather than a predefined "Refresh" label.

```python
proxy_features = [
    "search_volume",
    "competition",
    "competition_level",
    "content_type",
    "main_intent"
]

print("Target Type : Proxy")
print("Source      : Observed historical SEO data\n")

print("Available proxy-related features:")
for feature in proxy_features:
    print(f"- {feature}")
```

## 3. Success metric



The primary goal of this project is to help the SEO team identify the most valuable webpages to refresh first. Since only a limited number of pages can be updated at a time, the quality of the highest-ranked recommendations is more important than evaluating every page equally.

For this reason, **Precision@K** is the most appropriate success metric. It measures how many of the top **K** recommended pages are truly strong refresh candidates. A higher Precision@K means the model is successfully prioritizing the pages that are most likely to produce business value.

Metrics such as Accuracy are less suitable because they evaluate every page equally, while this project focuses on selecting the best pages rather than classifying the entire dataset.

```python
success_metric = "Precision@K"
top_k = 20

print(f"Chosen Metric : {success_metric}")
print(f"Evaluation    : Top {top_k} recommended webpages")
print("Goal          : Maximize the quality of refresh recommendations.")
```

## 5. Why ML beats a fixed rule here

The refresh opportunity decision cannot be captured by a simple if-statement because it depends on the interaction of multiple SEO features rather than a single condition.

For example, search volume alone is not enough to decide whether a page should be refreshed. The decision also depends on competition, competition level, content type, search intent, and other signals. The importance of one feature often changes depending on the values of the others.

Because these relationships are complex and vary across webpages, writing a fixed set of if-else rules would be difficult to maintain and would not generalize well. A Machine Learning model can learn these patterns from historical data and assign a more reliable refresh opportunity score.
